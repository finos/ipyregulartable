# ipyregulartable

Run the cells from top to bottom to create tables, handle clicks in Python, and apply styles.

```python
import ipywidgets as ipy

import ipyregulartable as rt
```

## Click events

Click a table cell to print its column and row coordinates below.

```python
click_table = rt.RegularTableWidget(rt.NumpyDataModel())
click_table
```

```python
click_output = ipy.Output()
click_output
```

```python
def on_click(widget, coords):
    with click_output:
        print(f"clicked: ({coords[0]},{coords[1]})")


click_table.on_click(on_click)
```

## Styling

Set the table height and style its headers from Python.

```python
import string
from random import choice, randint, random, sample

import numpy as np


def make_data(n_rows=100):
    return np.array(
        [
            (
                [_]
                + [
                    randint(0, 100),
                    "".join(sample(string.ascii_lowercase, 5)),
                    (random() * 100) - 50,
                    choice((True, False)),
                ]
            )
            for _ in range(n_rows)
        ]
    )
```

```python
styled_table = rt.RegularTableWidget(rt.NumpyDataModel(make_data()))
styled_table.height = 500
styled_table
```

```python
styled_table.css = {"th": "background-color: rgba(0, 0, 0, .5); color: white; font-weight: bold;"}
```

## pandas data models

Display a DataFrame or Series using the same widget.

```python
import pandas as pd

frame_table = rt.RegularTableWidget(pd.DataFrame({"symbol": ["AAPL", "MSFT"], "price": [201.5, 418.0]}))
frame_table
```

```python
series_table = rt.RegularTableWidget(pd.Series(["AAPL", "MSFT"], name="symbol"))
series_table
```
