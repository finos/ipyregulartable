```python
import ipyregulartable as rt
```

```python
import string
from random import choice, randint, random, sample

import numpy as np


def _generateRandomData(n_rows=100):
    return np.array(
        [
            (
                [_]
                + [
                    randint(0, 100),
                    "".join(sample(string.ascii_lowercase, 5)),
                    (random() * 100) - 50,
                    choice((True, False)),
                ]
            )
            for _ in range(n_rows)
        ]
    )
```

```python
w = rt.RegularTableWidget(rt.NumpyDataModel(_generateRandomData()))
w.height = 500
w
```

```python
w.css = {"th": "background-color: rgba(0, 0, 0, .5); color: white; text-weight: bold;"}
```
